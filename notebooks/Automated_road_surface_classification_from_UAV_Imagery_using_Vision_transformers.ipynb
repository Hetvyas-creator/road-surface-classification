```python


import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from transformers import SegformerForSemanticSegmentation
import torch.optim as optim

import albumentations as A
from albumentations.pytorch import ToTensorV2
import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from pathlib import Path
from tqdm.auto import tqdm
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings('ignore')

print("✅ Imports complete!")

# ============================================================================

class Config:
    # Paths
    DATA_DIR = Path("/content/drive/MyDrive/Drone_image")
    IMAGE_DIR = DATA_DIR / "images"
    MASK_DIR = DATA_DIR / "masks"
    OUTPUT_DIR = DATA_DIR / "output"

    # Model
    MODEL_NAME = "nvidia/segformer-b3-finetuned-ade-512-512"
    NUM_CLASSES = 6


    IMAGE_SIZE = 384       # Reduced from 512
    BATCH_SIZE = 1         # Reduced from 2
    NUM_EPOCHS = 40
    LEARNING_RATE = 5e-5
    WEIGHT_DECAY = 0.01

    # Data
    TRAIN_SIZE = 0.70
    NUM_AUGMENTATIONS = 8  # Reduced from 12

    # Device
    DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    # Classes
    CLASS_NAMES = {
        0: 'Background',
        1: 'Paver Blocks',
        2: 'RCC Road',
        3: 'Bitumen Road',
        4: 'Kachha Road',
        5: 'Vegetation'
    }

    CLASS_COLORS = {
        0: [0, 0, 0],
        1: [0, 255, 0],
        2: [0, 0, 255],
        3: [0, 255, 255],
        4: [255, 165, 0],
        5: [255, 255, 0]
    }

config = Config()
print(f"✅ Config set - Device: {config.DEVICE}")

# Clear GPU memory
import torch
torch.cuda.empty_cache()

# ============================================================================
# AUGMENTATION
# ============================================================================

def get_training_augmentation():
    return A.Compose([
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.3),
        A.RandomRotate90(p=0.3),
        A.ShiftScaleRotate(shift_limit=0.1, scale_limit=0.2, rotate_limit=30,
                          border_mode=cv2.BORDER_CONSTANT, value=0, mask_value=0, p=0.7),
        A.RandomBrightnessContrast(brightness_limit=0.3, contrast_limit=0.3, p=0.8),
        A.HueSaturationValue(hue_shift_limit=10, sat_shift_limit=30, val_shift_limit=30, p=0.7),
        A.RandomGamma(gamma_limit=(80, 120), p=0.5),
        A.OneOf([
            A.GaussNoise(var_limit=(10.0, 50.0)),
            A.GaussianBlur(blur_limit=(3, 7)),
            A.MotionBlur(blur_limit=5),
        ], p=0.5),
        A.RandomShadow(shadow_roi=(0, 0, 1, 1), num_shadows_lower=1,
                      num_shadows_upper=3, shadow_dimension=5, p=0.3),
        A.CoarseDropout(max_holes=8, max_height=32, max_width=32,
                       min_holes=2, min_height=8, min_width=8,
                       fill_value=0, mask_fill_value=0, p=0.3),
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ToTensorV2(),
    ])

def get_validation_augmentation():
    return A.Compose([
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ToTensorV2(),
    ])

print("✅ Augmentation pipelines ready")

# ============================================================================
# DATASET
# ============================================================================

class RoadDataset(Dataset):
    def __init__(self, image_paths, mask_paths, transform=None, augment_count=1):
        self.image_paths = image_paths
        self.mask_paths = mask_paths
        self.transform = transform
        self.augment_count = augment_count

        self.samples = []
        for img_path, mask_path in zip(image_paths, mask_paths):
            for aug_idx in range(augment_count):
                self.samples.append({'image': img_path, 'mask': mask_path, 'aug_idx': aug_idx})

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        sample = self.samples[idx]

        image = cv2.imread(str(sample['image']))
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        mask = np.array(Image.open(sample['mask']))

        if self.transform:
            transformed = self.transform(image=image, mask=mask)
            image = transformed['image']
            mask = transformed['mask']

        return {'pixel_values': image, 'labels': mask.long()}

# Prepare data
print("\n" + "="*80)
print("PREPARING DATA")
print("="*80)

image_files = sorted(list(config.IMAGE_DIR.glob("*.JPG")) + list(config.IMAGE_DIR.glob("*.jpg")))
mask_files = sorted(list(config.MASK_DIR.glob("*.png")))

valid_pairs = []
for img_path in image_files:
    mask_path = config.MASK_DIR / (img_path.stem + '.png')
    if mask_path.exists():
        valid_pairs.append((img_path, mask_path))

print(f"✅ {len(valid_pairs)} valid image-mask pairs")

train_pairs, val_pairs = train_test_split(valid_pairs, train_size=config.TRAIN_SIZE, random_state=42)

print(f"Split: {len(train_pairs)} train / {len(val_pairs)} val")

train_dataset = RoadDataset(
    [p[0] for p in train_pairs], [p[1] for p in train_pairs],
    transform=get_training_augmentation(), augment_count=config.NUM_AUGMENTATIONS
)

val_dataset = RoadDataset(
    [p[0] for p in val_pairs], [p[1] for p in val_pairs],
    transform=get_validation_augmentation(), augment_count=1
)

print(f"Train: {len(train_dataset)} samples ({len(train_pairs)} × {config.NUM_AUGMENTATIONS})")
print(f"Val: {len(val_dataset)} samples")

train_loader = DataLoader(train_dataset, batch_size=config.BATCH_SIZE, shuffle=True, num_workers=0, pin_memory=False)
val_loader = DataLoader(val_dataset, batch_size=config.BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=False)

print("✅ Dataloaders ready")

# ============================================================================
# MODEL
# ============================================================================

print("\n" + "="*80)
print("LOADING MODEL")
print("="*80)

model = SegformerForSemanticSegmentation.from_pretrained(
    config.MODEL_NAME,
    num_labels=config.NUM_CLASSES,
    ignore_mismatched_sizes=True
)

model = model.to(config.DEVICE)

total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"✅ Model loaded: SegFormer-B3")
print(f"  Total params: {total_params:,}")
print(f"  Trainable: {trainable_params:,}")



def calculate_iou(pred, target, num_classes):
    ious = []
    pred = pred.cpu().numpy()
    target = target.cpu().numpy()

    for cls in range(num_classes):
        pred_cls = (pred == cls)
        target_cls = (target == cls)
        intersection = (pred_cls & target_cls).sum()
        union = (pred_cls | target_cls).sum()
        iou = intersection / union if union > 0 else 0.0
        ious.append(iou)

    return np.array(ious)

def train_epoch(model, loader, criterion, optimizer):
    model.train()
    total_loss = 0
    all_ious = []

    pbar = tqdm(loader, desc="Training")
    for batch in pbar:
        pixel_values = batch['pixel_values'].to(config.DEVICE)
        labels = batch['labels'].to(config.DEVICE)

        outputs = model(pixel_values=pixel_values)
        logits = outputs.logits

        upsampled_logits = F.interpolate(logits, size=labels.shape[-2:],
                                        mode='bilinear', align_corners=False)

        loss = criterion(upsampled_logits, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        preds = upsampled_logits.argmax(dim=1)
        ious = calculate_iou(preds, labels, config.NUM_CLASSES)
        all_ious.append(ious)

        pbar.set_postfix({'loss': f'{loss.item():.4f}'})

    avg_loss = total_loss / len(loader)
    avg_iou = np.mean(all_ious, axis=0)
    mean_iou = np.mean(avg_iou)

    return avg_loss, mean_iou, avg_iou

def validate_epoch(model, loader, criterion):
    model.eval()
    total_loss = 0
    all_ious = []

    pbar = tqdm(loader, desc="Validation")
    with torch.no_grad():
        for batch in pbar:
            pixel_values = batch['pixel_values'].to(config.DEVICE)
            labels = batch['labels'].to(config.DEVICE)

            outputs = model(pixel_values=pixel_values)
            logits = outputs.logits

            upsampled_logits = F.interpolate(logits, size=labels.shape[-2:],
                                            mode='bilinear', align_corners=False)

            loss = criterion(upsampled_logits, labels)
            total_loss += loss.item()

            preds = upsampled_logits.argmax(dim=1)
            ious = calculate_iou(preds, labels, config.NUM_CLASSES)
            all_ious.append(ious)

            pbar.set_postfix({'loss': f'{loss.item():.4f}'})

    avg_loss = total_loss / len(loader)
    avg_iou = np.mean(all_ious, axis=0)
    mean_iou = np.mean(avg_iou)

    return avg_loss, mean_iou, avg_iou

# Train
print("\n" + "="*80)
print("STARTING TRAINING")
print("="*80 + "\n")

criterion = nn.CrossEntropyLoss(ignore_index=255)
optimizer = optim.AdamW(model.parameters(), lr=config.LEARNING_RATE, weight_decay=config.WEIGHT_DECAY)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=config.NUM_EPOCHS)

history = {'train_loss': [], 'train_iou': [], 'val_loss': [], 'val_iou': []}
best_val_iou = 0.0
patience = 10
patience_counter = 0

for epoch in range(config.NUM_EPOCHS):
    print(f"\nEpoch {epoch+1}/{config.NUM_EPOCHS}")
    print("-" * 80)

    train_loss, train_iou, train_class_iou = train_epoch(model, train_loader, criterion, optimizer)
    val_loss, val_iou, val_class_iou = validate_epoch(model, val_loader, criterion)

    scheduler.step()

    history['train_loss'].append(train_loss)
    history['train_iou'].append(train_iou)
    history['val_loss'].append(val_loss)
    history['val_iou'].append(val_iou)

    print(f"\nResults:")
    print(f"  Train Loss: {train_loss:.4f}, Train IoU: {train_iou:.4f}")
    print(f"  Val Loss:   {val_loss:.4f}, Val IoU:   {val_iou:.4f}")

    print(f"\nPer-class Val IoU:")
    for cls, iou in enumerate(val_class_iou):
        print(f"  {config.CLASS_NAMES[cls]:15s}: {iou:.4f}")

    if val_iou > best_val_iou:
        best_val_iou = val_iou
        patience_counter = 0

        config.OUTPUT_DIR.mkdir(exist_ok=True, parents=True)
        torch.save(model.state_dict(), config.OUTPUT_DIR / "best_model.pth")
        print(f"\n✅ Saved best model (Val IoU: {val_iou:.4f})")
    else:
        patience_counter += 1

    if patience_counter >= patience:
        print(f"\n⚠️  Early stopping at epoch {epoch+1}")
        break

print(f"\n{'='*80}")
print(f"TRAINING COMPLETE!")
print(f"{'='*80}")
print(f"Best Validation IoU: {best_val_iou:.4f}")

# ============================================================================
# VISUALIZATION
# ============================================================================

print("\n" + "="*80)
print("CREATING VISUALIZATIONS")
print("="*80 + "\n")

model.eval()

fig, axes = plt.subplots(5, 4, figsize=(20, 25))

with torch.no_grad():
    for idx, batch in enumerate(val_loader):
        if idx >= 5:
            break

        pixel_values = batch['pixel_values'].to(config.DEVICE)
        labels = batch['labels']

        outputs = model(pixel_values=pixel_values)
        logits = outputs.logits

        upsampled_logits = F.interpolate(logits, size=labels.shape[-2:],
                                        mode='bilinear', align_corners=False)
        preds = upsampled_logits.argmax(dim=1)

        img = pixel_values[0].cpu().permute(1, 2, 0).numpy()
        img = img * np.array([0.229, 0.224, 0.225]) + np.array([0.485, 0.456, 0.406])
        img = np.clip(img, 0, 1)

        mask_true = labels[0].cpu().numpy()
        mask_pred = preds[0].cpu().numpy()

        def colorize(mask):
            colored = np.zeros((*mask.shape, 3), dtype=np.uint8)
            for cls, color in config.CLASS_COLORS.items():
                colored[mask == cls] = color
            return colored

        mask_true_colored = colorize(mask_true)
        mask_pred_colored = colorize(mask_pred)
        overlay = (img * 0.6 + mask_pred_colored / 255 * 0.4)

        axes[idx, 0].imshow(img)
        axes[idx, 0].set_title('Original', fontsize=12)
        axes[idx, 0].axis('off')

        axes[idx, 1].imshow(mask_true_colored)
        axes[idx, 1].set_title('Ground Truth', fontsize=12)
        axes[idx, 1].axis('off')

        axes[idx, 2].imshow(mask_pred_colored)
        axes[idx, 2].set_title('Prediction', fontsize=12)
        axes[idx, 2].axis('off')

        axes[idx, 3].imshow(overlay)
        axes[idx, 3].set_title('Overlay', fontsize=12)
        axes[idx, 3].axis('off')

plt.tight_layout()
plt.savefig(config.OUTPUT_DIR / "predictions.png", dpi=150, bbox_inches='tight')
plt.show()

print(f"✅ Visualizations saved!")
print(f"\n{'='*80}")
print(f"ALL DONE!")
print(f"{'='*80}")
print(f"\nModel saved to: {config.OUTPUT_DIR / 'best_model2.pth'}")
print(f"Predictions saved to: {config.OUTPUT_DIR / 'predictions.png'}")
print(f"\nBest Validation IoU: {best_val_iou:.4f}")
```

```python
# ============================================================================
# FIXED INFERENCE WITH CORRECT MEASUREMENTS
# ============================================================================

# ... (keep all the imports and Config class same as before)

import torch
import torch.nn.functional as F
from transformers import SegformerForSemanticSegmentation
import albumentations as A
from albumentations.pytorch import ToTensorV2
import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from pathlib import Path
from skimage import measure, morphology
from skimage.measure import regionprops

# ============================================================================
# CONFIGURATION
# ============================================================================

class Config:
    MODEL_PATH = "/content/drive/MyDrive/Drone_image/output/best_model.pth"
    MODEL_NAME = "nvidia/segformer-b3-finetuned-ade-512-512"
    NUM_CLASSES = 6
    IMAGE_SIZE = 384
    DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    CLASS_NAMES = {
        0: 'Background',
        1: 'Paver Blocks',
        2: 'RCC Road',
        3: 'Bitumen Road',
        4: 'Kachha Road',
        5: 'Vegetation'
    }

    CLASS_COLORS = {
        0: [0, 0, 0],
        1: [0, 255, 0],
        2: [0, 0, 255],
        3: [0, 255, 255],
        4: [255, 165, 0],
        5: [255, 255, 0]
    }

    # Drone specs
    DRONE_ALTITUDE = 60.3              # relative altitude (NOT 89)
    SENSOR_WIDTH_MM = 9.6              # correct for FC3411
    FOCAL_LENGTH_MM = 8.38             # from metadata
    IMAGE_WIDTH_PIXELS = 5472

    gsd = (DRONE_ALTITUDE * SENSOR_WIDTH_MM) / (FOCAL_LENGTH_MM * IMAGE_WIDTH_PIXELS)

config = Config()

# ============================================================================
# FIXED: GSD CALCULATION BASED ON ACTUAL IMAGE SIZE
# ============================================================================

def calculate_gsd_for_image(altitude_meters, image_width_pixels):
    """
    Calculate GSD based on ACTUAL image width

    CRITICAL: Must use the actual image width, not model input size!
    """
    gsd_m = (config.SENSOR_WIDTH_MM * altitude_meters) / (
        config.FOCAL_LENGTH_MM * image_width_pixels
    )
    return gsd_m * 100  # cm per pixel

def get_measurements_for_class(mask, class_id, gsd):
    """Get area measurement for a specific class"""

    binary_mask = (mask == class_id).astype(np.uint8)
    binary_mask = morphology.remove_small_objects(
        binary_mask.astype(bool),
        min_size=100  # Increased from 50
    ).astype(np.uint8)

    if binary_mask.sum() == 0:
        return None

    labeled = measure.label(binary_mask, connectivity=2)
    regions = regionprops(labeled)

    if len(regions) == 0:
        return None

    # Get largest region
    largest = max(regions, key=lambda r: r.area)

    # Calculate area
    area_m2 = largest.area * (gsd / 100) ** 2

    centroid_y, centroid_x = largest.centroid

    return {
        'area_m2': area_m2,
        'centroid': (int(centroid_x), int(centroid_y))
    }

# ============================================================================
# FIXED: BETTER TEXT VISIBILITY
# ============================================================================

def add_measurements_to_overlay(overlay_image, mask, altitude=89):
    """
    Add measurement text with BETTER VISIBILITY
    """

    output = overlay_image.copy()

    # FIXED: Calculate GSD based on actual image width
    actual_width = overlay_image.shape[1]
    gsd = calculate_gsd_for_image(altitude, actual_width)

    print(f"\n📏 GSD Calculation:")
    print(f"   Image width: {actual_width} pixels")
    print(f"   GSD: {gsd:.3f} cm/pixel")

    # BETTER text settings for visibility
    font = cv2.FONT_HERSHEY_DUPLEX  # Changed font
    font_scale = 1.0  # Larger (was 0.7)
    thickness = 3  # Thicker (was 2)

    for class_id, class_name in config.CLASS_NAMES.items():

        if class_id == 0:
            continue

        measurement = get_measurements_for_class(mask, class_id, gsd)

        if measurement is None:
            continue

        # Create text
        text = f"{class_name}: {measurement['area_m2']:.1f}m²"

        # Get text size
        (tw, th), baseline = cv2.getTextSize(text, font, font_scale, thickness)

        # Position at centroid
        cx, cy = measurement['centroid']
        x = max(15, min(cx - tw // 2, output.shape[1] - tw - 15))
        y = max(th + 15, min(cy, output.shape[0] - 15))

        # BETTER background (larger, more opaque)
        padding = 12  # Increased from 6

        # Semi-transparent black background
        overlay_box = output.copy()
        cv2.rectangle(
            overlay_box,
            (x - padding, y - th - padding - baseline),
            (x + tw + padding, y + padding),
            (0, 0, 0),
            -1
        )
        cv2.addWeighted(overlay_box, 0.7, output, 0.3, 0, output)

        # White border (thicker)
        cv2.rectangle(
            output,
            (x - padding, y - th - padding - baseline),
            (x + tw + padding, y + padding),
            (255, 255, 255),
            3  # Thicker border
        )

        # Black outline for text (for even better visibility)
        cv2.putText(
            output,
            text,
            (x, y),
            font,
            font_scale,
            (0, 0, 0),  # Black outline
            thickness + 2  # Thicker
        )

        # White text on top
        cv2.putText(
            output,
            text,
            (x, y),
            font,
            font_scale,
            (255, 255, 255),
            thickness
        )

    return output

# ============================================================================
# PRINT MEASUREMENTS (FIXED)
# ============================================================================

def print_measurements(prediction_mask, altitude=60.3, image_width=None):
    """Print area measurements with CORRECT GSD"""

    if image_width is None:
        image_width = prediction_mask.shape[1]

    gsd = calculate_gsd_for_image(altitude, image_width)

    print("\n" + "="*80)
    print("AREA MEASUREMENTS (CORRECTED)")
    print("="*80)
    print(f"\nImage width: {image_width} pixels")
    print(f"Ground Sampling Distance: {gsd:.3f} cm/pixel")
    print(f"Altitude: {altitude} meters\n")

    total_area = 0

    for class_id, class_name in config.CLASS_NAMES.items():
        if class_id == 0:
            continue

        measurement = get_measurements_for_class(prediction_mask, class_id, gsd)

        if measurement:
            print(f"{class_name}:")
            print(f"  Area: {measurement['area_m2']:.2f} m²")
            total_area += measurement['area_m2']

    print(f"\nTotal classified area: {total_area:.2f} m²")
    print("="*80)

# ============================================================================
# KEEP ALL OTHER FUNCTIONS SAME (load_model, preprocess, predict, etc.)
# ============================================================================

def load_trained_model(model_path):
    """Load the trained model"""
    print("\n" + "="*80)
    print("LOADING TRAINED MODEL")
    print("="*80)

    model = SegformerForSemanticSegmentation.from_pretrained(
        config.MODEL_NAME,
        num_labels=config.NUM_CLASSES,
        ignore_mismatched_sizes=True
    )

    state_dict = torch.load(model_path, map_location=config.DEVICE)
    model.load_state_dict(state_dict)

    model = model.to(config.DEVICE)
    model.eval()

    print(f"✅ Model loaded from: {model_path}")
    print(f"✅ Device: {config.DEVICE}")

    return model

def get_inference_transform():
    return A.Compose([
        A.Resize(config.IMAGE_SIZE, config.IMAGE_SIZE),
        A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ToTensorV2(),
    ])

def preprocess_image(image_path):
    image = cv2.imread(str(image_path))
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    original_size = image.shape[:2]

    transform = get_inference_transform()
    transformed = transform(image=image)
    image_tensor = transformed['image'].unsqueeze(0)

    return image_tensor, original_size, image

def predict(model, image_tensor, original_size):
    with torch.no_grad():
        image_tensor = image_tensor.to(config.DEVICE)
        outputs = model(pixel_values=image_tensor)
        logits = outputs.logits

        upsampled_logits = F.interpolate(
            logits,
            size=original_size,
            mode='bilinear',
            align_corners=False
        )

        predictions = upsampled_logits.argmax(dim=1)[0].cpu().numpy()

    return predictions

def colorize_mask(mask):
    colored = np.zeros((*mask.shape, 3), dtype=np.uint8)
    for class_id, color in config.CLASS_COLORS.items():
        colored[mask == class_id] = color
    return colored

def calculate_statistics(prediction_mask):
    total_pixels = prediction_mask.size
    stats = {}

    for class_id, class_name in config.CLASS_NAMES.items():
        count = np.sum(prediction_mask == class_id)
        percentage = (count / total_pixels) * 100
        stats[class_name] = {
            'pixels': int(count),
            'percentage': float(percentage)
        }

    return stats

def print_statistics(stats):
    print("\n" + "="*80)
    print("DETECTION STATISTICS")
    print("="*80)

    sorted_stats = sorted(stats.items(), key=lambda x: x[1]['percentage'], reverse=True)

    print(f"\n{'Class':<20} {'Pixels':<15} {'Percentage':<10}")
    print("-" * 80)

    for class_name, data in sorted_stats:
        if data['percentage'] > 0.01:
            print(f"{class_name:<20} {data['pixels']:<15,} {data['percentage']:>6.2f}%")

    print("-" * 80)

def visualize_prediction(original_image, prediction_mask, altitude=89, save_path=None):
    """Create visualization with 4 panels"""

    colored_mask = colorize_mask(prediction_mask)
    overlay = cv2.addWeighted(original_image, 0.6, colored_mask, 0.4, 0)

    # FIXED: Pass actual image width for correct GSD
    overlay_with_measurements = add_measurements_to_overlay(
        overlay.copy(),
        prediction_mask,
        altitude
    )

    fig, axes = plt.subplots(2, 2, figsize=(20, 16))
    axes = axes.flatten()

    axes[0].imshow(original_image)
    axes[0].set_title('Original Image', fontsize=16, fontweight='bold')
    axes[0].axis('off')

    axes[1].imshow(colored_mask)
    axes[1].set_title('Predicted Segmentation', fontsize=16, fontweight='bold')
    axes[1].axis('off')

    axes[2].imshow(overlay)
    axes[2].set_title('Overlay', fontsize=16, fontweight='bold')
    axes[2].axis('off')

    axes[3].imshow(overlay_with_measurements)
    axes[3].set_title('Overlay + Measurements', fontsize=16, fontweight='bold')
    axes[3].axis('off')

    legend_elements = []
    for class_id, class_name in config.CLASS_NAMES.items():
        if class_id > 0:
            color = np.array(config.CLASS_COLORS[class_id]) / 255.0
            legend_elements.append(
                plt.Rectangle((0, 0), 1, 1, fc=color, label=class_name)
            )

    axes[1].legend(handles=legend_elements, loc='upper right', fontsize=9)
    axes[3].legend(handles=legend_elements, loc='upper right', fontsize=9)

    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"✅ Visualization saved: {save_path}")

    plt.show()

    return overlay_with_measurements



def run_inference(image_path, model_path=None, altitude=60.3, save_output=True):
    """
    FIXED: Now returns correct measurements and better visibility
    """

    print("\n" + "="*80)
    print("VISION TRANSFORMER INFERENCE WITH MEASUREMENTS")
    print("="*80)
    print(f"\nInput image: {image_path}")
    print(f"Altitude: {altitude} meters")

    if model_path is None:
        model_path = config.MODEL_PATH

    model = load_trained_model(model_path)

    print("\n→ Preprocessing image...")
    image_tensor, original_size, original_image = preprocess_image(image_path)
    print(f"✅ Original image size: {original_size[1]} × {original_size[0]} pixels")

    print("\n→ Running inference...")
    prediction_mask = predict(model, image_tensor, original_size)
    print(f"✅ Prediction complete!")

    stats = calculate_statistics(prediction_mask)
    print_statistics(stats)

    # FIXED: Pass image width for correct GSD
    print_measurements(prediction_mask, altitude, image_width=original_size[1])

    print("\n→ Creating visualization with measurements...")

    output_path = None
    if save_output:
        output_path = Path(image_path).parent / f"{Path(image_path).stem}_prediction.jpg"

    overlay_measured = visualize_prediction(
        original_image,
        prediction_mask,
        altitude,
        save_path=output_path
    )

    if save_output:
        measured_path = Path(image_path).parent / f"{Path(image_path).stem}_measured.jpg"
        cv2.imwrite(
            str(measured_path),
            cv2.cvtColor(overlay_measured, cv2.COLOR_RGB2BGR)
        )
        print(f"✅ Measured overlay saved: {measured_path}")

    print("\n" + "="*80)
    print("INFERENCE COMPLETE!")
    print("="*80)

    return prediction_mask, stats, overlay_measured  # Returns 3 values!



def run_batch_inference(image_folder, model_path=None, altitude=60.3, output_folder=None):
    """Batch processing with FIXED measurements"""

    image_folder = Path(image_folder)

    if output_folder is None:
        output_folder = image_folder / "predictions"
    output_folder = Path(output_folder)
    output_folder.mkdir(exist_ok=True, parents=True)

    if model_path is None:
        model_path = config.MODEL_PATH

    model = load_trained_model(model_path)

    image_files = list(image_folder.glob("*.jpg")) + list(image_folder.glob("*.JPG"))

    print(f"\nFound {len(image_files)} images to process")
    print("="*80)

    for i, image_path in enumerate(image_files, 1):
        print(f"\n[{i}/{len(image_files)}] Processing: {image_path.name}")

        image_tensor, original_size, original_image = preprocess_image(image_path)
        prediction_mask = predict(model, image_tensor, original_size)

        colored_mask = colorize_mask(prediction_mask)
        overlay = cv2.addWeighted(original_image, 0.6, colored_mask, 0.4, 0)

        overlay_measured = add_measurements_to_overlay(
            overlay.copy(),
            prediction_mask,
            altitude
        )

        measured_path = output_folder / f"{image_path.stem}_measured.jpg"
        cv2.imwrite(str(measured_path), cv2.cvtColor(overlay_measured, cv2.COLOR_RGB2BGR))

        print(f"  ✅ Saved: {measured_path.name}")

    print("\n" + "="*80)
    print(f"✅ Batch complete! Results in: {output_folder}")
    print("="*80)


if __name__ == "__main__":
    print("""
    FIXED VERSION - Correct measurements + Better visibility!

    Usage:
    prediction, stats, measured = run_inference("test.jpg", altitude=60.3)
    """)
```

```python
# ============================================================================
# COMPLETE INFRASTRUCTURE ASSESSMENT PIPELINE
# Combines: condition_assessment.py + priority_ranking.py +
#           report_generator.py + complete_pipeline.py
# ============================================================================

import numpy as np
import pandas as pd
from typing import Dict, List, Tuple
from datetime import datetime, timedelta
import json
import os

from reportlab.lib.pagesizes import A4
from reportlab.lib import colors
from reportlab.lib.units import inch
from reportlab.platypus import (SimpleDocTemplate, Table, TableStyle,
                                Paragraph, Spacer, PageBreak, Image)
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER, TA_LEFT, TA_JUSTIFY
import matplotlib.pyplot as plt


# ============================================================================
# MODULE 1: ROAD CONDITION ASSESSMENT ENGINE
# ============================================================================

class RoadConditionAssessor:
    """
    Converts segmentation results into actionable road condition scores
    """

    def __init__(self):
        # Surface quality base scores (0-100)
        self.SURFACE_QUALITY = {
            'Background': 0,      # Not road surface
            'Paver_Blocks': 95,   # Excellent - durable, aesthetic
            'RCC': 90,            # Good - long-lasting, smooth
            'Bitumen': 85,        # Good - standard for traffic
            'Kachha': 40,         # Poor - unpaved, dusty
            'Vegetation': 0       # Obstruction - needs clearing
        }

        # Condition thresholds
        self.CONDITION_LEVELS = {
            'Excellent': (85, 100),
            'Good': (70, 84),
            'Fair': (50, 69),
            'Poor': (30, 49),
            'Critical': (0, 29)
        }

        # Maintenance recommendations
        self.RECOMMENDATIONS = {
            'Excellent': 'Routine monitoring only',
            'Good': 'Preventive maintenance recommended',
            'Fair': 'Minor repairs needed within 6 months',
            'Poor': 'Significant repairs needed within 3 months',
            'Critical': 'Immediate intervention required'
        }

    def calculate_condition_score(self, measurements: Dict[str, Dict]) -> Dict:
        """
        Calculate overall road condition score

        Args:
            measurements: {
                'Bitumen': {'area_m2': 234.6, ...},
                'RCC': {'area_m2': 89.7, ...},
                ...
            }

        Returns:
            {
                'overall_score': 78.5,
                'condition_category': 'Good',
                'recommendation': 'Preventive maintenance recommended',
                'surface_breakdown': {...},
                'dominant_surface': 'Bitumen'
            }
        """

        # Calculate total road area (exclude background and vegetation)
        road_surfaces = ['Paver_Blocks', 'RCC', 'Bitumen', 'Kachha']
        total_road_area = sum(
            measurements.get(surf, {}).get('area_m2', 0)
            for surf in road_surfaces
        )

        if total_road_area == 0:
            return self._no_road_result()

        # Calculate weighted score
        weighted_sum = 0
        surface_breakdown = {}

        for surface_type, data in measurements.items():
            if surface_type in self.SURFACE_QUALITY:
                area = data.get('area_m2', 0)
                quality = self.SURFACE_QUALITY[surface_type]

                if surface_type in road_surfaces:
                    contribution = (area / total_road_area) * quality
                    weighted_sum += contribution

                    surface_breakdown[surface_type] = {
                        'area_m2': round(area, 2),
                        'percentage': round((area / total_road_area) * 100, 2),
                        'quality_score': quality,
                        'contribution_to_score': round(contribution, 2)
                    }

        overall_score = round(weighted_sum, 1)

        # Determine condition category
        condition_category = self._get_condition_category(overall_score)

        # Find dominant surface
        dominant_surface = max(
            surface_breakdown.items(),
            key=lambda x: x[1]['area_m2']
        )[0] if surface_breakdown else 'None'

        return {
            'overall_score': overall_score,
            'condition_category': condition_category,
            'recommendation': self.RECOMMENDATIONS[condition_category],
            'total_road_area_m2': round(total_road_area, 2),
            'surface_breakdown': surface_breakdown,
            'dominant_surface': dominant_surface,
            'assessment_metadata': {
                'scoring_method': 'Area-weighted quality averaging',
                'max_possible_score': 100,
                'min_possible_score': 0
            }
        }

    def _get_condition_category(self, score: float) -> str:
        """Determine condition category from score"""
        for category, (min_score, max_score) in self.CONDITION_LEVELS.items():
            if min_score <= score <= max_score:
                return category
        return 'Unknown'

    def _no_road_result(self) -> Dict:
        """Return result when no road detected"""
        return {
            'overall_score': 0,
            'condition_category': 'No Road Detected',
            'recommendation': 'No road surface identified in image',
            'total_road_area_m2': 0,
            'surface_breakdown': {},
            'dominant_surface': 'None'
        }

    def assess_multiple_images(self, all_measurements: Dict[str, Dict]) -> Dict:
        """
        Assess condition across multiple images

        Args:
            all_measurements: {
                'image_001': {measurements},
                'image_002': {measurements},
                ...
            }

        Returns:
            Aggregated assessment with statistics
        """

        results = {}
        all_scores = []

        for image_id, measurements in all_measurements.items():
            assessment = self.calculate_condition_score(measurements)
            results[image_id] = assessment

            if assessment['overall_score'] > 0:
                all_scores.append(assessment['overall_score'])

        # Calculate network-level statistics
        if all_scores:
            network_stats = {
                'average_score': round(np.mean(all_scores), 1),
                'median_score': round(np.median(all_scores), 1),
                'min_score': round(min(all_scores), 1),
                'max_score': round(max(all_scores), 1),
                'std_deviation': round(np.std(all_scores), 1),
                'total_images_assessed': len(all_scores)
            }

            # Distribution across categories
            category_counts = {}
            for result in results.values():
                cat = result['condition_category']
                category_counts[cat] = category_counts.get(cat, 0) + 1

            network_stats['category_distribution'] = category_counts
        else:
            network_stats = None

        return {
            'individual_assessments': results,
            'network_statistics': network_stats
        }


# ============================================================================
# MODULE 2: MAINTENANCE PRIORITIZATION SYSTEM
# ============================================================================

class MaintenancePrioritizer:
    """
    Ranks roads for maintenance intervention based on multiple factors
    """

    def __init__(self):
        # Factor weights (sum to 1.0)
        self.WEIGHTS = {
            'condition_score': 0.40,    # Condition is most important
            'road_importance': 0.30,    # Strategic value
            'traffic_volume': 0.20,     # Usage intensity
            'maintenance_history': 0.10 # Time since last repair
        }

        # Road importance multipliers
        self.IMPORTANCE_LEVELS = {
            'highway': 1.0,
            'main_road': 1.0,
            'arterial': 0.9,
            'residential': 0.7,
            'service': 0.5,
            'pedestrian': 0.6
        }

        # Traffic volume multipliers
        self.TRAFFIC_LEVELS = {
            'high': 1.0,      # >10,000 vehicles/day
            'medium': 0.7,    # 5,000-10,000
            'low': 0.4        # <5,000
        }

        # Repair cost estimates (₹ per m²)
        self.REPAIR_COSTS = {
            'Paver_Blocks': 450,
            'RCC': 550,
            'Bitumen': 500,
            'Kachha': 400  # Full paving cost
        }

    def calculate_priority_score(self,
                                 condition_assessment: Dict,
                                 road_metadata: Dict = None) -> Dict:
        """
        Calculate priority score for a single road segment

        Args:
            condition_assessment: output from Module 1
            road_metadata: {
                'road_type': 'main_road',
                'traffic_volume': 'high',
                'last_maintenance_date': '2019-03-15',
                'location': 'Campus Main Entrance',
                'length_m': 450
            }

        Returns:
            Priority scoring with recommendations
        """

        # Default metadata if not provided
        if road_metadata is None:
            road_metadata = {
                'road_type': 'residential',
                'traffic_volume': 'medium',
                'last_maintenance_date': None,
                'location': 'Unknown',
                'length_m': 100
            }

        # Factor 1: Condition Score (inverted - lower is worse, higher priority)
        condition_score = condition_assessment['overall_score']
        condition_factor = (100 - condition_score) / 100  # 0-1, higher is worse

        # Factor 2: Road Importance
        road_type = road_metadata.get('road_type', 'residential')
        importance_factor = self.IMPORTANCE_LEVELS.get(road_type, 0.5)

        # Factor 3: Traffic Volume
        traffic = road_metadata.get('traffic_volume', 'medium')
        traffic_factor = self.TRAFFIC_LEVELS.get(traffic, 0.7)

        # Factor 4: Maintenance History
        maintenance_factor = self._calculate_maintenance_factor(
            road_metadata.get('last_maintenance_date')
        )

        # Calculate weighted priority score (0-100)
        priority_score = (
            condition_factor * self.WEIGHTS['condition_score'] +
            importance_factor * self.WEIGHTS['road_importance'] +
            traffic_factor * self.WEIGHTS['traffic_volume'] +
            maintenance_factor * self.WEIGHTS['maintenance_history']
        ) * 100

        # Determine urgency level
        urgency = self._get_urgency_level(priority_score)

        # Estimate repair cost
        repair_cost = self._estimate_repair_cost(condition_assessment)

        # Recommended timeline
        timeline = self._recommend_timeline(urgency, condition_score)

        return {
            'priority_score': round(priority_score, 1),
            'urgency_level': urgency,
            'condition_score': condition_score,
            'estimated_cost_inr': repair_cost,
            'recommended_timeline': timeline,
            'factors': {
                'condition_factor': round(condition_factor, 3),
                'importance_factor': importance_factor,
                'traffic_factor': traffic_factor,
                'maintenance_factor': round(maintenance_factor, 3)
            },
            'metadata': road_metadata
        }

    def rank_roads(self,
                   assessments: Dict[str, Dict],
                   metadata_lookup: Dict[str, Dict] = None) -> pd.DataFrame:
        """
        Rank multiple roads by priority

        Args:
            assessments: {image_id: condition_assessment}
            metadata_lookup: {image_id: road_metadata}

        Returns:
            DataFrame sorted by priority (highest first)
        """

        results = []

        for image_id, assessment in assessments.items():
            metadata = metadata_lookup.get(image_id, {}) if metadata_lookup else {}

            priority_result = self.calculate_priority_score(assessment, metadata)

            results.append({
                'image_id': image_id,
                'location': metadata.get('location', image_id),
                'priority_score': priority_result['priority_score'],
                'urgency_level': priority_result['urgency_level'],
                'condition_score': priority_result['condition_score'],
                'condition_category': assessment['condition_category'],
                'estimated_cost_inr': priority_result['estimated_cost_inr'],
                'timeline': priority_result['recommended_timeline'],
                'road_type': metadata.get('road_type', 'Unknown'),
                'total_area_m2': assessment['total_road_area_m2']
            })

        # Create DataFrame and sort by priority
        df = pd.DataFrame(results)
        df = df.sort_values('priority_score', ascending=False).reset_index(drop=True)
        df['rank'] = df.index + 1

        return df

    def create_maintenance_schedule(self,
                                    ranked_roads: pd.DataFrame,
                                    annual_budget_inr: float) -> Dict:
        """
        Generate maintenance schedule within budget constraints

        Args:
            ranked_roads: output from rank_roads()
            annual_budget_inr: available budget (₹)

        Returns:
            Phased maintenance plan
        """

        # Phase 1: Critical (immediate)
        critical = ranked_roads[ranked_roads['urgency_level'] == 'Critical']

        # Phase 2: High (within 3 months)
        high = ranked_roads[ranked_roads['urgency_level'] == 'High']

        # Phase 3: Medium (within 6 months)
        medium = ranked_roads[ranked_roads['urgency_level'] == 'Medium']

        # Calculate cumulative costs
        phases = {
            'Phase_1_Critical': {
                'roads': critical.to_dict('records'),
                'total_cost': critical['estimated_cost_inr'].sum(),
                'timeline': 'Immediate (0-1 month)'
            },
            'Phase_2_High': {
                'roads': high.to_dict('records'),
                'total_cost': high['estimated_cost_inr'].sum(),
                'timeline': 'Short-term (1-3 months)'
            },
            'Phase_3_Medium': {
                'roads': medium.to_dict('records'),
                'total_cost': medium['estimated_cost_inr'].sum(),
                'timeline': 'Medium-term (3-6 months)'
            }
        }

        total_required = sum(p['total_cost'] for p in phases.values())

        budget_status = {
            'annual_budget_inr': annual_budget_inr,
            'total_required_inr': total_required,
            'budget_deficit_inr': max(0, total_required - annual_budget_inr),
            'budget_surplus_inr': max(0, annual_budget_inr - total_required),
            'can_complete_all': total_required <= annual_budget_inr
        }

        return {
            'phases': phases,
            'budget_analysis': budget_status,
            'recommendations': self._generate_budget_recommendations(
                phases, budget_status
            )
        }

    def _calculate_maintenance_factor(self, last_maintenance_date: str) -> float:
        """Calculate factor based on time since last maintenance"""
        if last_maintenance_date is None:
            return 1.0  # Never maintained - highest priority

        try:
            last_date = datetime.strptime(last_maintenance_date, '%Y-%m-%d')
            years_since = (datetime.now() - last_date).days / 365.25

            if years_since > 5:
                return 0.8
            elif years_since > 3:
                return 0.6
            else:
                return 0.3
        except:
            return 0.5  # Default if date parsing fails

    def _get_urgency_level(self, priority_score: float) -> str:
        """Determine urgency level from priority score"""
        if priority_score >= 75:
            return 'Critical'
        elif priority_score >= 60:
            return 'High'
        elif priority_score >= 40:
            return 'Medium'
        else:
            return 'Low'

    def _estimate_repair_cost(self, condition_assessment: Dict) -> float:
        """Estimate total repair cost based on surface areas"""
        total_cost = 0
        breakdown = condition_assessment.get('surface_breakdown', {})

        # Cost poor quality surfaces
        poor_surfaces = ['Kachha', 'Vegetation']

        for surface, data in breakdown.items():
            if surface in poor_surfaces or condition_assessment['overall_score'] < 50:
                area = data.get('area_m2', 0)
                cost_per_m2 = self.REPAIR_COSTS.get(surface, 500)
                total_cost += area * cost_per_m2

        return round(total_cost, 2)

    def _recommend_timeline(self, urgency: str, condition_score: float) -> str:
        """Recommend maintenance timeline"""
        timelines = {
            'Critical': 'Immediate action required (0-1 month)',
            'High': 'Short-term intervention needed (1-3 months)',
            'Medium': 'Schedule within 6 months',
            'Low': 'Routine maintenance cycle (annual)'
        }
        return timelines.get(urgency, 'Monitor regularly')

    def _generate_budget_recommendations(self, phases: Dict, budget: Dict) -> List[str]:
        """Generate recommendations based on budget analysis"""
        recommendations = []

        if budget['can_complete_all']:
            recommendations.append(
                f"Budget sufficient: ₹{budget['budget_surplus_inr']:,.0f} surplus available"
            )
            recommendations.append("Recommend completing all phases as scheduled")
        else:
            deficit = budget['budget_deficit_inr']
            recommendations.append(
                f"Budget shortfall: ₹{deficit:,.0f} additional funding needed"
            )
            recommendations.append(
                "Priority: Complete Critical phase immediately"
            )
            recommendations.append(
                "Defer Medium priority repairs to next fiscal year"
            )

        return recommendations


# ============================================================================
# MODULE 3: INFRASTRUCTURE REPORT GENERATOR
# ============================================================================

class InfrastructureReportGenerator:
    """
    Generates professional PDF reports for stakeholders
    """

    def __init__(self, output_dir='./reports'):
        self.output_dir = output_dir
        os.makedirs(output_dir, exist_ok=True)
        self.styles = getSampleStyleSheet()
        self._setup_custom_styles()

    def _setup_custom_styles(self):
        """Create custom paragraph styles"""
        self.styles.add(ParagraphStyle(
            name='CustomTitle',
            parent=self.styles['Heading1'],
            fontSize=24,
            textColor=colors.HexColor('#1f4788'),
            spaceAfter=30,
            alignment=TA_CENTER
        ))

        self.styles.add(ParagraphStyle(
            name='SectionHeader',
            parent=self.styles['Heading2'],
            fontSize=16,
            textColor=colors.HexColor('#2c5aa0'),
            spaceAfter=12,
            spaceBefore=12
        ))

    def generate_full_report(self,
                             network_stats: Dict,
                             ranked_roads: pd.DataFrame,
                             maintenance_schedule: Dict,
                             project_info: Dict = None) -> str:
        """
        Generate complete infrastructure assessment report

        Args:
            network_stats: from Module 1 (network-level statistics)
            ranked_roads: from Module 2 (priority ranking)
            maintenance_schedule: from Module 2 (phased plan)
            project_info: {
                'survey_date': '2026-04-17',
                'location': 'Indrashil University Campus',
                'surveyed_by': 'Het Vyas',
                'total_area_km2': 0.26
            }

        Returns:
            Path to generated PDF
        """

        # Default project info
        if project_info is None:
            project_info = {
                'survey_date': datetime.now().strftime('%Y-%m-%d'),
                'location': 'Survey Area',
                'surveyed_by': 'Automated System',
                'total_area_km2': 0.5
            }

        # Create PDF filename
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        filename = f"infrastructure_report_{timestamp}.pdf"
        filepath = os.path.join(self.output_dir, filename)

        # Create document
        doc = SimpleDocTemplate(filepath, pagesize=A4,
                                rightMargin=0.75*inch, leftMargin=0.75*inch,
                                topMargin=1*inch, bottomMargin=1*inch)

        # Build content
        story = []

        # Title page
        story.extend(self._create_title_page(project_info))
        story.append(PageBreak())

        # Executive summary
        story.extend(self._create_executive_summary(network_stats, ranked_roads))
        story.append(Spacer(1, 0.3*inch))

        # Network overview
        story.extend(self._create_network_overview(network_stats))
        story.append(Spacer(1, 0.3*inch))

        # Priority rankings
        story.extend(self._create_priority_section(ranked_roads))
        story.append(PageBreak())

        # Maintenance schedule
        story.extend(self._create_maintenance_schedule(maintenance_schedule))
        story.append(Spacer(1, 0.3*inch))

        # Budget analysis
        story.extend(self._create_budget_analysis(maintenance_schedule))
        story.append(Spacer(1, 0.3*inch))

        # Recommendations
        story.extend(self._create_recommendations(maintenance_schedule))

        # Build PDF
        doc.build(story)

        print(f"✅ Report generated: {filepath}")
        return filepath

    def _create_title_page(self, project_info: Dict) -> List:
        """Create title page elements"""
        elements = []

        # Title
        title = Paragraph(
            "ROAD INFRASTRUCTURE<br/>CONDITION ASSESSMENT REPORT",
            self.styles['CustomTitle']
        )
        elements.append(Spacer(1, 2*inch))
        elements.append(title)
        elements.append(Spacer(1, 0.5*inch))

        # Project details
        details = f"""
        <para alignment='center' fontSize='12'>
        <b>Survey Location:</b> {project_info['location']}<br/>
        <b>Survey Date:</b> {project_info['survey_date']}<br/>
        <b>Total Area Surveyed:</b> {project_info['total_area_km2']} km²<br/>
        <b>Surveyed By:</b> {project_info['surveyed_by']}<br/>
        <b>Report Generated:</b> {datetime.now().strftime('%Y-%m-%d %H:%M')}
        </para>
        """
        elements.append(Paragraph(details, self.styles['Normal']))

        return elements

    def _create_executive_summary(self, network_stats: Dict, ranked_roads: pd.DataFrame) -> List:
        """Create executive summary section"""
        elements = []

        elements.append(Paragraph("EXECUTIVE SUMMARY", self.styles['SectionHeader']))

        # Key findings
        avg_score = network_stats['average_score']
        critical_count = len(ranked_roads[ranked_roads['urgency_level'] == 'Critical'])
        total_cost = ranked_roads['estimated_cost_inr'].sum()

        summary_text = f"""
        <para alignment='justify'>
        This report presents the results of an automated road surface condition assessment
        conducted using UAV imagery and computer vision analysis. The survey evaluated
        {network_stats['total_images_assessed']} road segments covering the designated area.
        <br/><br/>
        <b>Key Findings:</b><br/>
        • Average Network Condition Score: <b>{avg_score}/100</b><br/>
        • Road segments requiring immediate attention: <b>{critical_count}</b><br/>
        • Estimated total repair cost: <b>₹{total_cost:,.0f}</b><br/>
        • Overall network condition: <b>{self._get_network_assessment(avg_score)}</b>
        </para>
        """
        elements.append(Paragraph(summary_text, self.styles['Normal']))

        return elements

    def _create_network_overview(self, network_stats: Dict) -> List:
        """Create network statistics overview"""
        elements = []

        elements.append(Paragraph("NETWORK CONDITION OVERVIEW", self.styles['SectionHeader']))

        # Statistics table
        stats_data = [
            ['Metric', 'Value'],
            ['Average Condition Score', f"{network_stats['average_score']}/100"],
            ['Median Condition Score', f"{network_stats['median_score']}/100"],
            ['Best Segment Score', f"{network_stats['max_score']}/100"],
            ['Worst Segment Score', f"{network_stats['min_score']}/100"],
            ['Standard Deviation', f"{network_stats['std_deviation']}"],
            ['Total Segments Assessed', f"{network_stats['total_images_assessed']}"]
        ]

        table = Table(stats_data, colWidths=[3*inch, 2*inch])
        table.setStyle(TableStyle([
            ('BACKGROUND', (0, 0), (-1, 0), colors.grey),
            ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
            ('ALIGN', (0, 0), (-1, -1), 'LEFT'),
            ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
            ('FONTSIZE', (0, 0), (-1, 0), 12),
            ('BOTTOMPADDING', (0, 0), (-1, 0), 12),
            ('BACKGROUND', (0, 1), (-1, -1), colors.beige),
            ('GRID', (0, 0), (-1, -1), 1, colors.black)
        ]))

        elements.append(table)

        # Category distribution
        if 'category_distribution' in network_stats:
            elements.append(Spacer(1, 0.2*inch))
            elements.append(Paragraph("Condition Category Distribution:", self.styles['Heading3']))

            cat_data = [['Category', 'Count', 'Percentage']]
            total = network_stats['total_images_assessed']

            for category, count in network_stats['category_distribution'].items():
                percentage = (count / total) * 100
                cat_data.append([category, str(count), f"{percentage:.1f}%"])

            cat_table = Table(cat_data, colWidths=[2*inch, 1.5*inch, 1.5*inch])
            cat_table.setStyle(TableStyle([
                ('BACKGROUND', (0, 0), (-1, 0), colors.grey),
                ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
                ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
                ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
                ('GRID', (0, 0), (-1, -1), 1, colors.black),
                ('BACKGROUND', (0, 1), (-1, -1), colors.lightgrey)
            ]))

            elements.append(cat_table)

        return elements

    def _create_priority_section(self, ranked_roads: pd.DataFrame) -> List:
        """Create priority ranking section"""
        elements = []

        elements.append(Paragraph("MAINTENANCE PRIORITY RANKINGS", self.styles['SectionHeader']))

        # Top 10 priority roads
        top_roads = ranked_roads.head(10)

        table_data = [['Rank', 'Location', 'Urgency', 'Condition', 'Est. Cost (₹)']]

        for _, row in top_roads.iterrows():
            table_data.append([
                str(int(row['rank'])),
                row['location'][:30],  # Truncate long names
                row['urgency_level'],
                f"{row['condition_score']}/100",
                f"{row['estimated_cost_inr']:,.0f}"
            ])

        priority_table = Table(table_data, colWidths=[0.6*inch, 2.2*inch, 1*inch, 1*inch, 1.2*inch])
        priority_table.setStyle(TableStyle([
            ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#1f4788')),
            ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
            ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
            ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
            ('FONTSIZE', (0, 0), (-1, 0), 10),
            ('BOTTOMPADDING', (0, 0), (-1, 0), 12),
            ('GRID', (0, 0), (-1, -1), 1, colors.black),
            ('ROWBACKGROUNDS', (0, 1), (-1, -1), [colors.white, colors.lightgrey])
        ]))

        elements.append(priority_table)

        return elements

    def _create_maintenance_schedule(self, schedule: Dict) -> List:
        """Create maintenance schedule section"""
        elements = []

        elements.append(Paragraph("RECOMMENDED MAINTENANCE SCHEDULE", self.styles['SectionHeader']))

        phases = schedule['phases']

        for phase_name, phase_data in phases.items():
            phase_title = phase_name.replace('_', ' ').title()
            elements.append(Paragraph(f"<b>{phase_title}</b>", self.styles['Heading3']))

            phase_text = f"""
            <para>
            Timeline: {phase_data['timeline']}<br/>
            Number of Segments: {len(phase_data['roads'])}<br/>
            Estimated Cost: ₹{phase_data['total_cost']:,.0f}
            </para>
            """
            elements.append(Paragraph(phase_text, self.styles['Normal']))
            elements.append(Spacer(1, 0.15*inch))

        return elements

    def _create_budget_analysis(self, schedule: Dict) -> List:
        """Create budget analysis section"""
        elements = []

        elements.append(Paragraph("BUDGET ANALYSIS", self.styles['SectionHeader']))

        budget = schedule['budget_analysis']

        budget_data = [
            ['Budget Item', 'Amount (₹)'],
            ['Available Annual Budget', f"{budget['annual_budget_inr']:,.0f}"],
            ['Total Required Funding', f"{budget['total_required_inr']:,.0f}"],
            ['Budget Surplus / (Deficit)',
             f"{budget.get('budget_surplus_inr', 0) - budget.get('budget_deficit_inr', 0):,.0f}"]
        ]

        budget_table = Table(budget_data, colWidths=[3*inch, 2*inch])
        budget_table.setStyle(TableStyle([
            ('BACKGROUND', (0, 0), (-1, 0), colors.grey),
            ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
            ('ALIGN', (0, 0), (-1, -1), 'LEFT'),
            ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
            ('GRID', (0, 0), (-1, -1), 1, colors.black),
            ('BACKGROUND', (0, 1), (-1, -1), colors.lightblue)
        ]))

        elements.append(budget_table)

        return elements

    def _create_recommendations(self, schedule: Dict) -> List:
        """Create recommendations section"""
        elements = []

        elements.append(Paragraph("RECOMMENDATIONS", self.styles['SectionHeader']))

        recommendations = schedule.get('recommendations', [])

        rec_text = "<para alignment='justify'>"
        for i, rec in enumerate(recommendations, 1):
            rec_text += f"{i}. {rec}<br/>"
        rec_text += "</para>"

        elements.append(Paragraph(rec_text, self.styles['Normal']))

        return elements

    def _get_network_assessment(self, avg_score: float) -> str:
        """Get overall network assessment"""
        if avg_score >= 85:
            return "Excellent - Minimal intervention required"
        elif avg_score >= 70:
            return "Good - Preventive maintenance recommended"
        elif avg_score >= 50:
            return "Fair - Active maintenance program needed"
        elif avg_score >= 30:
            return "Poor - Significant repairs required"
        else:
            return "Critical - Immediate comprehensive intervention necessary"


# ============================================================================
# INTEGRATION SCRIPT: Ties all modules together
# ============================================================================

def run_complete_assessment_pipeline(measurements_dict: Dict,
                                     metadata_dict: Dict = None,
                                     annual_budget: float = 1000000) -> Dict:
    """
    Complete end-to-end assessment pipeline

    Args:
        measurements_dict: {image_id: {surface: {area_m2: X}}}
        metadata_dict: {image_id: {road_type, traffic_volume, etc.}}
        annual_budget: Available budget in ₹

    Returns:
        Complete analysis results + PDF report path
    """

    print("=" * 80)
    print("RUNNING COMPLETE INFRASTRUCTURE ASSESSMENT PIPELINE")
    print("=" * 80)

    # Step 1: Condition Assessment
    print("\n[1/4] Assessing road conditions...")
    assessor = RoadConditionAssessor()
    network_assessment = assessor.assess_multiple_images(measurements_dict)

    individual_assessments = network_assessment['individual_assessments']
    network_stats = network_assessment['network_statistics']

    print(f"  ✓ Assessed {len(individual_assessments)} road segments")
    print(f"  ✓ Network average score: {network_stats['average_score']}/100")

    # Step 2: Priority Ranking
    print("\n[2/4] Ranking roads by maintenance priority...")
    prioritizer = MaintenancePrioritizer()
    ranked_roads = prioritizer.rank_roads(individual_assessments, metadata_dict)

    critical_count = len(ranked_roads[ranked_roads['urgency_level'] == 'Critical'])
    print(f"  ✓ Identified {critical_count} critical priority roads")

    # Step 3: Maintenance Scheduling
    print("\n[3/4] Creating maintenance schedule...")
    maintenance_schedule = prioritizer.create_maintenance_schedule(
        ranked_roads,
        annual_budget
    )

    total_cost = ranked_roads['estimated_cost_inr'].sum()
    print(f"  ✓ Total estimated cost: ₹{total_cost:,.0f}")
    print(f"  ✓ Budget: ₹{annual_budget:,.0f}")

    # Step 4: Report Generation
    print("\n[4/4] Generating infrastructure report...")
    report_gen = InfrastructureReportGenerator()

    project_info = {
        'survey_date': datetime.now().strftime('%Y-%m-%d'),
        'location': 'Indrashil University Campus',
        'surveyed_by': 'Automated UAV Assessment System',
        'total_area_km2': 0.26
    }

    report_path = report_gen.generate_full_report(
        network_stats,
        ranked_roads,
        maintenance_schedule,
        project_info
    )

    print(f"\n{'=' * 80}")
    print("ASSESSMENT COMPLETE!")
    print(f"{'=' * 80}")
    print(f"Report saved to: {report_path}")

    return {
        'network_stats': network_stats,
        'individual_assessments': individual_assessments,
        'ranked_roads': ranked_roads,
        'maintenance_schedule': maintenance_schedule,
        'report_path': report_path
    }


# ============================================================================
# USAGE EXAMPLE
# ============================================================================

if __name__ == "__main__":
    # Example measurements from your existing pipeline
    example_measurements = {
        'img_001': {
            'Bitumen': {'area_m2': 150.0},
            'Kachha': {'area_m2': 50.0}
        },
        'img_002': {
            'RCC': {'area_m2': 200.0},
            'Paver_Blocks': {'area_m2': 80.0}
        },
        'img_003': {
            'Bitumen': {'area_m2': 234.6},
            'RCC': {'area_m2': 89.7},
            'Paver_Blocks': {'area_m2': 45.2},
            'Kachha': {'area_m2': 12.4},
            'Vegetation': {'area_m2': 8.1}
        }
    }

    metadata = {
        'img_001': {
            'road_type': 'main_road',
            'traffic_volume': 'high',
            'location': 'Main Gate Entry',
            'last_maintenance_date': '2019-03-15'
        },
        'img_002': {
            'road_type': 'pedestrian',
            'traffic_volume': 'low',
            'location': 'Garden Path'
        },
        'img_003': {
            'road_type': 'arterial',
            'traffic_volume': 'medium',
            'location': 'Campus Ring Road'
        }
    }

    results = run_complete_assessment_pipeline(
        measurements_dict=example_measurements,
        metadata_dict=metadata,
        annual_budget=1000000
    )
```

```python
# ============================================================================
# AUTOMATED ROAD SURFACE CLASSIFICATION - COMPLETE INFERENCE PIPELINE
# ============================================================================

import torch
import numpy as np
from PIL import Image
from transformers import SegformerImageProcessor, SegformerForSemanticSegmentation
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from google.colab.patches import cv2_imshow
import cv2

# ============================================================================
# CONFIGURATION
# ============================================================================

CLASS_NAMES = ['Background', 'Paver Blocks', 'RCC', 'Bitumen', 'Kachha', 'Vegetation']
CLASS_COLORS = {
    0: [0, 0, 0],         # Background - Black
    1: [0, 255, 0],       # Paver Blocks - Green
    2: [0, 0, 255],       # RCC - Blue
    3: [0, 255, 255],     # Bitumen - Cyan
    4: [255, 165, 0],     # Kachha - Orange
    5: [255, 255, 0]      # Vegetation - Yellow
}

# Camera specifications (DJI Phantom 4 Pro)
SENSOR_WIDTH_MM = 13.2
FOCAL_LENGTH_MM = 8.8
IMAGE_WIDTH_PX = 5472

# Surface quality scores (0-100)
SURFACE_QUALITY = {
    'Background': 0,
    'Paver Blocks': 95,
    'RCC': 90,
    'Bitumen': 85,
    'Kachha': 40,
    'Vegetation': 0
}

# Repair costs (₹ per m²)
REPAIR_COSTS = {
    'Background': 0,
    'Paver Blocks': 450,
    'RCC': 550,
    'Bitumen': 500,
    'Kachha': 400,
    'Vegetation': 0
}

# ============================================================================
# LOAD MODEL
# ============================================================================

def load_model(checkpoint_path):
    """Load trained SegFormer model"""
    print("🔧 Loading model...")

    processor = SegformerImageProcessor.from_pretrained(
        "nvidia/segformer-b3-finetuned-ade-512-512"
    )

    model = SegformerForSemanticSegmentation.from_pretrained(
        "nvidia/segformer-b3-finetuned-ade-512-512",
        num_labels=len(CLASS_NAMES),
        id2label={i: name for i, name in enumerate(CLASS_NAMES)},
        label2id={name: i for i, name in enumerate(CLASS_NAMES)},
        ignore_mismatched_sizes=True
    )

    # Load trained weights - FIXED
    checkpoint = torch.load(checkpoint_path, map_location='cpu')

    # Try different checkpoint formats
    if isinstance(checkpoint, dict):
        if 'model_state_dict' in checkpoint:
            model.load_state_dict(checkpoint['model_state_dict'])
        elif 'state_dict' in checkpoint:
            model.load_state_dict(checkpoint['state_dict'])
        else:
            # Checkpoint is a dict but not wrapped - try loading directly
            model.load_state_dict(checkpoint)
    else:
        # Checkpoint is the state_dict itself
        model.load_state_dict(checkpoint)

    model.eval()

    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model.to(device)

    print(f"✅ Model loaded on {device}")
    return model, processor, device

# ============================================================================
# GSD CALCULATION
# ============================================================================

def calculate_gsd(altitude_m):
    """Calculate Ground Sampling Distance"""
    gsd_cm = (SENSOR_WIDTH_MM * altitude_m * 100) / (FOCAL_LENGTH_MM * IMAGE_WIDTH_PX)
    return gsd_cm

def extract_altitude_from_exif(image_path):
    """Extract altitude from EXIF metadata"""
    try:
        from PIL.ExifTags import TAGS
        image = Image.open(image_path)
        exif = image._getexif()

        if exif:
            for tag_id, value in exif.items():
                tag = TAGS.get(tag_id, tag_id)
                if tag == 'GPSInfo':
                    gps_info = value
                    if 6 in gps_info:  # GPSAltitude
                        altitude = float(gps_info[6])
                        return altitude
    except:
        pass

    # Default altitude if EXIF not available
    return 89.0  # meters (typical flight altitude)

# ============================================================================
# INFERENCE
# ============================================================================

def run_inference_core(image_path, model, processor, device):
    """Run segmentation inference"""
    print(f"\n🖼️  Processing: {image_path}")

    # Load and preprocess
    image = Image.open(image_path).convert('RGB')
    inputs = processor(images=image, return_tensors="pt")
    inputs = {k: v.to(device) for k, v in inputs.items()}

    # Inference
    with torch.no_grad():
        outputs = model(**inputs)
        logits = outputs.logits

    # Upsample to original size
    upsampled_logits = torch.nn.functional.interpolate(
        logits,
        size=image.size[::-1],
        mode='bilinear',
        align_corners=False
    )

    # Get predictions
    pred_seg = upsampled_logits.argmax(dim=1)[0].cpu().numpy()

    return np.array(image), pred_seg

# ============================================================================
# MEASUREMENT & STATISTICS
# ============================================================================

def calculate_measurements(pred_mask, altitude_m):
    """Calculate area measurements for each class"""
    gsd_cm = calculate_gsd(altitude_m)
    gsd_m = gsd_cm / 100.0
    pixel_area_m2 = gsd_m ** 2

    measurements = {}
    stats = {}

    total_pixels = pred_mask.size

    for class_id, class_name in enumerate(CLASS_NAMES):
        class_pixels = np.sum(pred_mask == class_id)
        area_m2 = class_pixels * pixel_area_m2
        percentage = (class_pixels / total_pixels) * 100

        measurements[class_name] = {
            'pixels': int(class_pixels),
            'area_m2': round(area_m2, 2),
            'percentage': round(percentage, 2)
        }

        stats[class_name] = {
            'area_m2': round(area_m2, 2),
            'percentage': round(percentage, 2)
        }

    return measurements, stats, gsd_cm

# ============================================================================
# DECISION LAYER: CONDITION ASSESSMENT
# ============================================================================

def assess_road_condition(measurements):
    """
    Convert measurements into actionable road condition assessment

    Returns:
        condition_score: 0-100 (higher is better)
        condition_category: Excellent/Good/Fair/Poor/Critical
        recommendation: Action needed
    """

    # Calculate total road area (exclude Background and Vegetation)
    road_surfaces = ['Paver Blocks', 'RCC', 'Bitumen', 'Kachha']
    total_road_area = sum(
        measurements[surf]['area_m2']
        for surf in road_surfaces
        if surf in measurements
    )

    if total_road_area == 0:
        return {
            'score': 0,
            'category': 'No Road Detected',
            'recommendation': 'No road surface identified',
            'repair_needed': False,
            'urgency': 'None'
        }

    # Calculate weighted condition score
    weighted_score = 0
    for surface_name, data in measurements.items():
        if surface_name in road_surfaces:
            area = data['area_m2']
            quality = SURFACE_QUALITY[surface_name]
            contribution = (area / total_road_area) * quality
            weighted_score += contribution

    score = round(weighted_score, 1)

    # Categorize condition
    if score >= 85:
        category = 'Excellent'
        recommendation = 'Routine monitoring only'
        urgency = 'Low'
        repair_needed = False
    elif score >= 70:
        category = 'Good'
        recommendation = 'Preventive maintenance recommended'
        urgency = 'Low'
        repair_needed = False
    elif score >= 50:
        category = 'Fair'
        recommendation = 'Minor repairs needed within 6 months'
        urgency = 'Medium'
        repair_needed = True
    elif score >= 30:
        category = 'Poor'
        recommendation = 'Significant repairs needed within 3 months'
        urgency = 'High'
        repair_needed = True
    else:
        category = 'Critical'
        recommendation = 'Immediate intervention required'
        urgency = 'Critical'
        repair_needed = True

    return {
        'score': score,
        'category': category,
        'recommendation': recommendation,
        'repair_needed': repair_needed,
        'urgency': urgency,
        'total_road_area_m2': round(total_road_area, 2)
    }

# ============================================================================
# DECISION LAYER: COST ESTIMATION
# ============================================================================

def estimate_repair_cost(measurements, condition_assessment):
    """
    Estimate repair costs based on surface conditions

    Returns:
        total_cost: Total estimated repair cost in ₹
        cost_breakdown: Cost per surface type
    """

    if not condition_assessment['repair_needed']:
        return {
            'total_cost_inr': 0,
            'cost_breakdown': {},
            'cost_per_m2_avg': 0,
            'notes': 'No repairs needed - road in good condition'
        }

    # Calculate costs for poor-quality surfaces
    poor_surfaces = ['Kachha', 'Vegetation']
    cost_breakdown = {}
    total_cost = 0

    for surface_name, data in measurements.items():
        area = data['area_m2']

        # Cost poor surfaces or deteriorated areas
        if surface_name in poor_surfaces or condition_assessment['score'] < 50:
            if area > 0:
                cost_per_m2 = REPAIR_COSTS[surface_name]
                surface_cost = area * cost_per_m2

                cost_breakdown[surface_name] = {
                    'area_m2': round(area, 2),
                    'cost_per_m2': cost_per_m2,
                    'total_cost': round(surface_cost, 2)
                }

                total_cost += surface_cost

    avg_cost_per_m2 = (
        total_cost / condition_assessment['total_road_area_m2']
        if condition_assessment['total_road_area_m2'] > 0
        else 0
    )

    return {
        'total_cost_inr': round(total_cost, 2),
        'cost_breakdown': cost_breakdown,
        'cost_per_m2_avg': round(avg_cost_per_m2, 2),
        'notes': f"Repair cost estimate for {condition_assessment['urgency']} priority maintenance"
    }

# ============================================================================
# DECISION LAYER: PRIORITY ASSESSMENT
# ============================================================================

def calculate_priority_score(condition_assessment, road_metadata=None):
    """
    Calculate maintenance priority score

    Args:
        condition_assessment: from assess_road_condition()
        road_metadata: Optional dict with road_type, traffic_volume, etc.

    Returns:
        Priority score and ranking
    """

    # Default metadata
    if road_metadata is None:
        road_metadata = {
            'road_type': 'campus_road',
            'traffic_volume': 'medium',
            'location_importance': 'medium'
        }

    # Base priority from condition (inverted - lower condition = higher priority)
    condition_priority = (100 - condition_assessment['score']) / 100

    # Road importance multiplier
    importance_weights = {
        'highway': 1.0,
        'main_road': 1.0,
        'campus_road': 0.8,
        'residential': 0.7,
        'service': 0.5
    }
    importance = importance_weights.get(road_metadata.get('road_type', 'campus_road'), 0.7)

    # Traffic multiplier
    traffic_weights = {
        'high': 1.0,
        'medium': 0.7,
        'low': 0.4
    }
    traffic = traffic_weights.get(road_metadata.get('traffic_volume', 'medium'), 0.7)

    # Calculate composite priority score
    priority_score = (
        condition_priority * 0.5 +
        importance * 0.3 +
        traffic * 0.2
    ) * 100

    # Determine priority rank
    if priority_score >= 75:
        priority_rank = 'Critical Priority'
        timeline = 'Immediate (0-1 month)'
    elif priority_score >= 60:
        priority_rank = 'High Priority'
        timeline = 'Short-term (1-3 months)'
    elif priority_score >= 40:
        priority_rank = 'Medium Priority'
        timeline = 'Medium-term (3-6 months)'
    else:
        priority_rank = 'Low Priority'
        timeline = 'Routine maintenance cycle'

    return {
        'priority_score': round(priority_score, 1),
        'priority_rank': priority_rank,
        'recommended_timeline': timeline,
        'factors': {
            'condition_factor': round(condition_priority, 2),
            'importance_factor': importance,
            'traffic_factor': traffic
        }
    }

# ============================================================================
# VISUALIZATION
# ============================================================================

def create_visualization(original_img, pred_mask, measurements,
                        condition_assessment, cost_estimate, priority):
    """Create comprehensive visualization with all insights"""

    # Create color mask
    h, w = pred_mask.shape
    color_mask = np.zeros((h, w, 3), dtype=np.uint8)

    for class_id, color in CLASS_COLORS.items():
        color_mask[pred_mask == class_id] = color

    # Create overlay
    overlay = cv2.addWeighted(original_img, 0.6, color_mask, 0.4, 0)

    # Add text annotations
    overlay_annotated = overlay.copy()

    # Title
    # Title
    cv2.putText(overlay_annotated, "ROAD CONDITION ASSESSMENT",
            (20, 40), cv2.FONT_HERSHEY_SIMPLEX, 1.2, (255, 255, 255), 3)

    # Condition Score (Top-left box)
    score = condition_assessment['score']
    category = condition_assessment['category']

    # Color based on score
    if score >= 70:
        score_color = (0, 255, 0)  # Green
    elif score >= 50:
        score_color = (0, 255, 255)  # Yellow
    else:
        score_color = (0, 0, 255)  # Red

    cv2.rectangle(overlay_annotated, (15, 60), (450, 180), (0, 0, 0), -1)
    cv2.rectangle(overlay_annotated, (15, 60), (450, 180), (255, 255, 255), 2)

    cv2.putText(overlay_annotated, f"Condition Score: {score}/100",
                (25, 95), cv2.FONT_HERSHEY_SIMPLEX, 0.8, score_color, 2)
    cv2.putText(overlay_annotated, f"Category: {category}",
                (25, 125), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
    cv2.putText(overlay_annotated, f"Priority: {priority['priority_rank']}",
                (25, 155), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 0), 2)

    # Cost & Timeline (Bottom-left box)
    y_start = h - 120
    cv2.rectangle(overlay_annotated, (15, y_start), (450, h - 15), (0, 0, 0), -1)
    cv2.rectangle(overlay_annotated, (15, y_start), (450, h - 15), (255, 255, 255), 2)

    cv2.putText(overlay_annotated, f"Repair Cost: Rs {cost_estimate['total_cost_inr']:,.0f}",
                (25, y_start + 35), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)
    cv2.putText(overlay_annotated, f"Timeline: {priority['recommended_timeline']}",
                (25, y_start + 65), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)
    cv2.putText(overlay_annotated, condition_assessment['recommendation'],
                (25, y_start + 90), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (200, 200, 200), 1)

    return color_mask, overlay_annotated

def display_results(original_img, color_mask, overlay_annotated, measurements,
                   condition_assessment, cost_estimate, priority):
    """Display all results in organized format"""

    # Create figure with subplots
    fig = plt.figure(figsize=(20, 12))

    # Original Image
    ax1 = plt.subplot(2, 3, 1)
    ax1.imshow(original_img)
    ax1.set_title('Original Image', fontsize=14, fontweight='bold')
    ax1.axis('off')

    # Prediction Mask
    ax2 = plt.subplot(2, 3, 2)
    ax2.imshow(color_mask)
    ax2.set_title('Segmentation Prediction', fontsize=14, fontweight='bold')
    ax2.axis('off')

    # Annotated Overlay
    ax3 = plt.subplot(2, 3, 3)
    ax3.imshow(overlay_annotated)
    ax3.set_title('Assessment Report', fontsize=14, fontweight='bold')
    ax3.axis('off')

    # Surface Distribution (Bar Chart)
    ax4 = plt.subplot(2, 3, 4)
    surfaces = []
    areas = []
    for name, data in measurements.items():
        if data['area_m2'] > 0 and name not in ['Background', 'Vegetation']:
            surfaces.append(name)
            areas.append(data['area_m2'])

    colors_bar = [np.array(CLASS_COLORS[CLASS_NAMES.index(s)])/255 for s in surfaces]
    ax4.bar(surfaces, areas, color=colors_bar)
    ax4.set_ylabel('Area (m²)', fontsize=12)
    ax4.set_title('Surface Distribution', fontsize=14, fontweight='bold')
    ax4.tick_params(axis='x', rotation=45)
    ax4.grid(axis='y', alpha=0.3)

    # Condition Assessment (Text Box)
    ax5 = plt.subplot(2, 3, 5)
    ax5.axis('off')

    assessment_text = f"""
    ╔══════════════════════════════════════╗
    ║     CONDITION ASSESSMENT REPORT      ║
    ╚══════════════════════════════════════╝

    Overall Score:        {condition_assessment['score']}/100
    Condition Category:   {condition_assessment['category']}
    Urgency Level:        {condition_assessment['urgency']}

    Total Road Area:      {condition_assessment['total_road_area_m2']} m²

    ╔══════════════════════════════════════╗
    ║      PRIORITY & COST ANALYSIS        ║
    ╚══════════════════════════════════════╝

    Priority Score:       {priority['priority_score']}/100
    Priority Rank:        {priority['priority_rank']}

    Estimated Cost:       ₹{cost_estimate['total_cost_inr']:,.0f}
    Avg Cost/m²:          ₹{cost_estimate['cost_per_m2_avg']:.2f}

    ╔══════════════════════════════════════╗
    ║         RECOMMENDATION               ║
    ╚══════════════════════════════════════╝

    {condition_assessment['recommendation']}

    Timeline: {priority['recommended_timeline']}
    """

    ax5.text(0.1, 0.5, assessment_text,
             fontsize=11, family='monospace',
             verticalalignment='center',
             bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.3))

    # Legend
    ax6 = plt.subplot(2, 3, 6)
    ax6.axis('off')

    legend_elements = [
        mpatches.Patch(color=np.array(color)/255, label=name)
        for name, color in zip(CLASS_NAMES, CLASS_COLORS.values())
    ]
    ax6.legend(handles=legend_elements, loc='center',
              fontsize=12, title='Surface Classes', title_fontsize=14)

    plt.tight_layout()
    plt.show()

    # Print detailed summary
    print("\n" + "="*80)
    print("📊 DETAILED ASSESSMENT SUMMARY")
    print("="*80)

    print(f"\n🏗️  ROAD CONDITION:")
    print(f"   Score: {condition_assessment['score']}/100")
    print(f"   Category: {condition_assessment['category']}")
    print(f"   Total Road Area: {condition_assessment['total_road_area_m2']} m²")

    print(f"\n⚠️  PRIORITY ASSESSMENT:")
    print(f"   Priority Score: {priority['priority_score']}/100")
    print(f"   Priority Rank: {priority['priority_rank']}")
    print(f"   Urgency: {condition_assessment['urgency']}")
    print(f"   Timeline: {priority['recommended_timeline']}")

    print(f"\n💰 COST ESTIMATION:")
    print(f"   Total Repair Cost: ₹{cost_estimate['total_cost_inr']:,.0f}")
    print(f"   Average Cost per m²: ₹{cost_estimate['cost_per_m2_avg']:.2f}")

    if cost_estimate['cost_breakdown']:
        print(f"\n   Cost Breakdown:")
        for surface, details in cost_estimate['cost_breakdown'].items():
            print(f"     • {surface}: {details['area_m2']} m² × ₹{details['cost_per_m2']} = ₹{details['total_cost']:,.2f}")

    print(f"\n📋 RECOMMENDATION:")
    print(f"   {condition_assessment['recommendation']}")

    print(f"\n📏 SURFACE MEASUREMENTS:")
    for name, data in measurements.items():
        if data['area_m2'] > 0:
            print(f"   • {name}: {data['area_m2']} m² ({data['percentage']}%)")

    print("\n" + "="*80)

# ============================================================================
# MAIN INFERENCE FUNCTION
# ============================================================================

def run_inference(image_path, model_checkpoint='best_model.pth', road_metadata=None):
    """
    Complete inference pipeline with decision support

    Args:
        image_path: Path to input image
        model_checkpoint: Path to trained model
        road_metadata: Optional metadata for priority calculation

    Returns:
        Complete assessment results
    """

    # Load model
    model, processor, device = load_model(model_checkpoint)

    # Extract altitude
    altitude_m = extract_altitude_from_exif(image_path)
    print(f"📐 Altitude: {altitude_m} m")
    print(f"📐 GSD: {calculate_gsd(altitude_m):.2f} cm/pixel")

    # Run inference
    original_img, pred_mask = run_inference_core(image_path, model, processor, device)

    # Calculate measurements
    measurements, stats, gsd_cm = calculate_measurements(pred_mask, altitude_m)
    print(f"✅ Measurements calculated (GSD: {gsd_cm:.2f} cm/pixel)")

    # === DECISION LAYER ===

    # 1. Assess road condition
    condition_assessment = assess_road_condition(measurements)
    print(f"\n✅ Condition Assessment: {condition_assessment['category']} ({condition_assessment['score']}/100)")

    # 2. Estimate repair costs
    cost_estimate = estimate_repair_cost(measurements, condition_assessment)
    print(f"✅ Cost Estimate: ₹{cost_estimate['total_cost_inr']:,.0f}")

    # 3. Calculate priority
    priority = calculate_priority_score(condition_assessment, road_metadata)
    print(f"✅ Priority: {priority['priority_rank']} (Score: {priority['priority_score']}/100)")

    # Create visualization
    color_mask, overlay_annotated = create_visualization(
        original_img, pred_mask, measurements,
        condition_assessment, cost_estimate, priority
    )

    # Display results
    display_results(
        original_img, color_mask, overlay_annotated, measurements,
        condition_assessment, cost_estimate, priority
    )

    # Return complete results
    return {
        'prediction_mask': pred_mask,
        'measurements': measurements,
        'stats': stats,
        'condition_assessment': condition_assessment,
        'cost_estimate': cost_estimate,
        'priority': priority,
        'gsd_cm': gsd_cm,
        'altitude_m': altitude_m
    }

# ============================================================================
# USAGE
# ============================================================================

# Upload image
# Upload image
from google.colab import files
print("📤 Click 'Choose Files' and select your drone image:")
uploaded = files.upload()

# Get filename
test_image = list(uploaded.keys())[0]
print(f"✅ Uploaded: {test_image}")

# Road metadata
road_metadata = {
    'road_type': 'campus_road',
    'traffic_volume': 'medium',
    'location_importance': 'medium'
}

# Run inference with correct model path
results = run_inference(test_image,
                       model_checkpoint=MODEL_PATH,  # Use the path variable
                       road_metadata=road_metadata)
```
